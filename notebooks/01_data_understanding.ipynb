### Missing and Duplicated Values Check

Missing and duplicated values can affect analysis quality. Therefore, the dataset must be checked for incomplete or duplicated reccords.

```python
df.isna().sum()
```

```python
str_columns = df.select_dtypes('str').columns
for col in str_columns:
    boolean_df = df[col].str.strip() == ""
    if boolean_df[1]:
        print('Empty String Found')
```

# Conclusion

The initial exploration provided a comprehensive understanding of the dataset structure and quality.

**Key findings:**

* The dataset contains **3,909 job postings** and **17 features** related to data science careers.
* No explicit missing values or duplicate records were identified.
* Several features use **`-1` as a placeholder value** instead of standard missing values. This was observed in `Rating`, `Founded`, `Industry`, `Competitors`, and `Easy Apply`.
* The `Competitors` and `Easy Apply` features contain a high proportion of placeholder values, which may limit their usefulness for further analysis.
* The `Job Title` feature contains many unique and highly specific job titles, indicating that role standardization may be beneficial for role-based analysis.
* The `Salary Estimate` feature is stored as text and will require preprocessing before salary analysis can be performed.

**Next Steps (Data Cleaning):**

* Remove unnecessary index columns (`Unnamed: 0` and `index`).
* Replace placeholder values (`-1`) with `NaN` where appropriate.
* Evaluate whether features with a high percentage of unavailable information should be retained or removed.
* Clean and transform the `Salary Estimate` feature into numerical variables.
* Standardize job titles by creating broader job categories while preserving the original titles.
* Save the cleaned dataset for subsequent exploratory data analysis.
